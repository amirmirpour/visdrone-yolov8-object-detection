# YOLOv8 Installation and Environment Test

**Project:** YOLOv8 Object Detection on VisDrone  
**Purpose:** Verify the Python environment, PyTorch, CUDA, GPU, and YOLOv8 model loading before starting experiments.

This notebook is used for environment verification only. Dataset preparation, training, and evaluation will be performed in separate notebooks.

```python
import sys
import torch
import ultralytics
from ultralytics import YOLO

print("Imports completed successfully.")
```

```python
print("Python version:", sys.version.split()[0])
print("Python executable:", sys.executable)
print("PyTorch version:", torch.__version__)
print("Ultralytics version:", ultralytics.__version__)
```

```python
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("CUDA build:", torch.version.cuda)
    print("GPU:", torch.cuda.get_device_name(0))
    print("GPU count:", torch.cuda.device_count())
else:
    print("GPU acceleration is unavailable.")
```

```python
model = YOLO("yolov8n.pt")

print("Model loaded successfully.")
print("Task:", model.task)
print("Model type:", type(model).__name__)
```

```python
environment_report = {
    "python_version": sys.version.split()[0],
    "pytorch_version": torch.__version__,
    "ultralytics_version": ultralytics.__version__,
    "cuda_available": torch.cuda.is_available(),
    "gpu_name": (
        torch.cuda.get_device_name(0)
        if torch.cuda.is_available()
        else "Not available"
    ),
    "model": "YOLOv8n",
    "task": model.task,
}

for key, value in environment_report.items():
    print(f"{key}: {value}")
```
